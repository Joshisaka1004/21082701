# 005 – Geometrie im Koordinatensystem

Jetzt bekommen die Figuren Koordinaten. Ein Punkt ist einfach ein Tupel `(x, y)`,
ein Vieleck eine Liste von Punkten.

Das ist genau die Stelle, an der Python richtig Spaß macht: Formeln, die man von Hand
nur ungern zweimal rechnet, sind als Funktion einmal geschrieben und dann fertig.

## Aufgabe 1 – Abstand und Mittelpunkt

Gegeben sind A(1 | 2) und B(7 | 10).

1. Länge der Strecke AB (Pythagoras im Koordinatensystem)
2. Mittelpunkt von AB

```python
import math

def abstand(P, Q):
    return math.sqrt((Q[0] - P[0])**2 + (Q[1] - P[1])**2)

def mittelpunkt(P, Q):
    return ((P[0] + Q[0]) / 2, (P[1] + Q[1]) / 2)

A = (1, 2)
B = (7, 10)

print(f"1) |AB| = {abstand(A, B)}")
mx, my = mittelpunkt(A, B)
print(f"2) M = ({mx} | {my})")
```

## Aufgabe 2 – Geraden

Eine Gerade schreiben wir als Paar `(m, b)` für `y = m·x + b`.

1. Bestimme die Gerade g durch A(1 | 2) und B(7 | 10).
2. Bestimme die Gerade h durch C(0 | 9) und D(6 | 0).
3. Schneide g und h.
4. Prüfe zwei weitere Geraden auf parallel bzw. senkrecht
   (senkrecht heißt: `m₁ · m₂ = −1`).

```python
def gerade_durch(P, Q):
    # liefert (m, b) fuer y = m*x + b
    if P[0] == Q[0]:
        raise ValueError("senkrechte Gerade - hier nicht als y = m*x + b darstellbar")
    m = (Q[1] - P[1]) / (Q[0] - P[0])
    b = P[1] - m * P[0]
    return m, b

def schnittpunkt(g, h):
    m1, b1 = g
    m2, b2 = h
    if math.isclose(m1, m2):
        return None                       # parallel -> kein einzelner Schnittpunkt
    x = (b2 - b1) / (m1 - m2)
    return (x, m1 * x + b1)

g = gerade_durch((1, 2), (7, 10))
h = gerade_durch((0, 9), (6, 0))
print(f"1) g: y = {g[0]:.4f}x + {g[1]:.4f}")
print(f"2) h: y = {h[0]:.4f}x + {h[1]:.4f}")

sx, sy = schnittpunkt(g, h)
print(f"3) Schnittpunkt S = ({sx:.4f} | {sy:.4f})")

for g1, g2, beschriftung in [((2, 1), (2, -5), "g1, g2"), ((2, 1), (-0.5, 3), "g1, g3")]:
    print(f"4) {beschriftung}: parallel = {math.isclose(g1[0], g2[0])}, "
          f"senkrecht = {math.isclose(g1[0] * g2[0], -1)}")
```

## Aufgabe 3 – Abstand eines Punktes von einer Geraden

Wie weit ist P(4 | 1) von der Geraden durch A(0 | 0) und B(6 | 8) entfernt?

Trick ohne Formelsammlung: Das Kreuzprodukt `(B−A) × (P−A)` ergibt die Fläche des
aufgespannten Parallelogramms. Teilt man sie durch die Grundseite `|AB|`, bleibt die Höhe –
und das ist genau der gesuchte Abstand.

```python
def abstand_punkt_gerade(P, A, B):
    flaeche = abs((B[0] - A[0]) * (P[1] - A[1]) - (B[1] - A[1]) * (P[0] - A[0]))
    return flaeche / abstand(A, B)

P = (4, 1)
A, B = (0, 0), (6, 8)
print(f"Abstand von P zur Geraden AB: {abstand_punkt_gerade(P, A, B)}")

# Probe: der Lotfusspunkt liegt auf AB und ist genau so weit entfernt
print("Punkt auf der Geraden (3 | 4):", abstand_punkt_gerade((3, 4), A, B))
```

## Aufgabe 4 – Fläche eines Vielecks (Schnürsenkelformel)

Ein Grundstück hat die Eckpunkte (0|0), (10|0), (12|6), (5|9), (−2|4).
Wie groß ist es?

Die **Gaußsche Trapezformel** (auch Schnürsenkelformel genannt) läuft einmal um das
Vieleck herum und summiert `x₁·y₂ − x₂·y₁` für jede Kante:

`A = ½ · |Σ (xᵢ·yᵢ₊₁ − xᵢ₊₁·yᵢ)|`

Zur Kontrolle rechnen wir ein Dreieck zusätzlich mit Heron aus Notebook 003.

```python
def polygon_flaeche(punkte):
    n = len(punkte)
    summe = 0
    for i in range(n):
        x1, y1 = punkte[i]
        x2, y2 = punkte[(i + 1) % n]      # nach dem letzten Punkt kommt wieder der erste
        summe += x1 * y2 - x2 * y1
    return abs(summe) / 2

def flaeche_heron(a, b, c):
    s = (a + b + c) / 2
    return math.sqrt(s * (s - a) * (s - b) * (s - c))

grundstueck = [(0, 0), (10, 0), (12, 6), (5, 9), (-2, 4)]
print(f"Flaeche des Fuenfecks: {polygon_flaeche(grundstueck)}")

dreieck = [(0, 0), (7, 0), (2, 5)]
seiten = [abstand(dreieck[0], dreieck[1]),
          abstand(dreieck[1], dreieck[2]),
          abstand(dreieck[2], dreieck[0])]
print(f"Dreieck, Schnuersenkel: {polygon_flaeche(dreieck)}")
print(f"Dreieck, Heron:         {flaeche_heron(*seiten):.4f}")
```

## Aufgabe 5 – Liegt der Punkt im Grundstück?

Klassische Frage aus jeder Landkarten-App. Die **Strahl-Methode**: Schicke von dem Punkt
aus einen Strahl nach rechts und zähle, wie oft er den Rand kreuzt.

* ungerade Anzahl → der Punkt liegt **innen**
* gerade Anzahl → der Punkt liegt **außen**

```python
def punkt_im_polygon(P, punkte):
    x, y = P
    drin = False
    n = len(punkte)
    for i in range(n):
        x1, y1 = punkte[i]
        x2, y2 = punkte[(i + 1) % n]
        # kreuzt die Kante die Hoehe y ueberhaupt?
        if (y1 > y) != (y2 > y):
            x_schnitt = x1 + (y - y1) * (x2 - x1) / (y2 - y1)
            if x < x_schnitt:              # Schnitt liegt rechts vom Punkt
                drin = not drin
    return drin

for P in [(5, 4), (11, 5), (0, 8), (13, 2)]:
    lage = "innerhalb" if punkt_im_polygon(P, grundstueck) else "ausserhalb"
    print(f"P{P}: {lage}")
```

## Aufgabe 6 – Merkwürdige Punkte im Dreieck

Für das Dreieck A(0|0), B(8|0), C(0|6):

1. Schwerpunkt (Mittelwert der drei Ecken)
2. Umkreismittelpunkt und Umkreisradius
3. Probe: Der Umkreismittelpunkt muss von allen drei Ecken gleich weit entfernt sein.

Bonus: Bei einem rechtwinkligen Dreieck liegt der Umkreismittelpunkt genau in der Mitte
der Hypotenuse – das ist der **Satz des Thales**. Sieht man das im Ergebnis?

```python
def schwerpunkt(A, B, C):
    return ((A[0] + B[0] + C[0]) / 3, (A[1] + B[1] + C[1]) / 3)

def umkreismittelpunkt(A, B, C):
    ax, ay = A
    bx, by = B
    cx, cy = C
    d = 2 * (ax * (by - cy) + bx * (cy - ay) + cx * (ay - by))
    ux = ((ax**2 + ay**2) * (by - cy) + (bx**2 + by**2) * (cy - ay)
          + (cx**2 + cy**2) * (ay - by)) / d
    uy = ((ax**2 + ay**2) * (cx - bx) + (bx**2 + by**2) * (ax - cx)
          + (cx**2 + cy**2) * (bx - ax)) / d
    return (ux, uy)

A, B, C = (0, 0), (8, 0), (0, 6)

sx, sy = schwerpunkt(A, B, C)
print(f"1) Schwerpunkt      S = ({sx:.4f} | {sy:.4f})")

U = umkreismittelpunkt(A, B, C)
print(f"2) Umkreismittelpunkt U = ({U[0]} | {U[1]})")
print(f"   Umkreisradius       r = {abstand(U, A)}")

print("3) Abstaende zu A, B, C:", [round(abstand(U, X), 4) for X in (A, B, C)])
print("   Mitte der Hypotenuse BC:", mittelpunkt(B, C), "-> Satz des Thales")
```
