## Aufgabe 1 – Satz des Pythagoras

Im rechtwinkligen Dreieck gilt `a² + b² = c²` (c ist die Hypotenuse, also die längste Seite).

1. Berechne die Hypotenuse für a = 3, b = 4.
2. Berechne die fehlende Kathete für c = 13, a = 5.
3. Prüfe, ob (7, 24, 25) und (6, 8, 11) rechtwinklige Dreiecke sind.

```python
import math

def hypotenuse(a, b):
    # c aus den beiden Katheten
    return math.sqrt(a**2 + b**2)

def kathete(c, a):
    # zweite Kathete aus Hypotenuse und einer Kathete
    return math.sqrt(c**2 - a**2)

def ist_rechtwinklig(a, b, c):
    # die laengste Seite muss die Hypotenuse sein -> erst sortieren
    a, b, c = sorted([a, b, c])
    return math.isclose(a**2 + b**2, c**2)

print("1) c =", hypotenuse(3, 4))
print("2) b =", kathete(13, 5))

for seiten in [(7, 24, 25), (6, 8, 11)]:
    antwort = "ja" if ist_rechtwinklig(*seiten) else "nein"
    print(f"3) {seiten}: rechtwinklig? {antwort}")
```

## Aufgabe 2 – Dreieck: Umfang, Fläche, Höhen

Ein Dreieck hat die Seiten a = 7, b = 8, c = 9.

1. Gibt es dieses Dreieck überhaupt? (Dreiecksungleichung: die Summe zweier Seiten
   muss größer als die dritte sein)
2. Umfang
3. Fläche mit der Formel von Heron: `s = (a+b+c)/2`, `A = √(s·(s−a)·(s−b)·(s−c))`
4. Die drei Höhen. Trick: Die Fläche ist immer `A = ½ · Seite · zugehörige Höhe`,
   also ist `h_a = 2A/a`.

```python
def existiert_dreieck(a, b, c):
    return a + b > c and a + c > b and b + c > a

def flaeche_heron(a, b, c):
    s = (a + b + c) / 2
    return math.sqrt(s * (s - a) * (s - b) * (s - c))

a, b, c = 7, 8, 9

print("1) Dreiecksungleichung erfuellt:", existiert_dreieck(a, b, c))
print("   Gegenbeispiel (1, 2, 9):    ", existiert_dreieck(1, 2, 9))

U = a + b + c
A = flaeche_heron(a, b, c)
print(f"2) Umfang  U = {U}")
print(f"3) Flaeche A = {A:.4f}")

for seite, name in [(a, "h_a"), (b, "h_b"), (c, "h_c")]:
    print(f"4) {name} = {2 * A / seite:.4f}")
```

## Aufgabe 3 – Kreis, Kreisring und Kreissektor

1. Kreis mit r = 5: Umfang und Fläche.
2. Kreisring mit R = 5 (außen) und r = 3 (innen): Flächeninhalt.
3. Kreissektor („Tortenstück") mit r = 6 und Mittelpunktswinkel α = 72°:
   Bogenlänge und Flächeninhalt.

Idee beim Sektor: Ein Sektor ist einfach der Bruchteil `α/360°` des ganzen Kreises.

```python
def kreis_umfang(r):
    return 2 * math.pi * r

def kreis_flaeche(r):
    return math.pi * r**2

def ring_flaeche(R, r):
    return kreis_flaeche(R) - kreis_flaeche(r)

def sektor(r, alpha_grad):
    anteil = alpha_grad / 360          # welcher Bruchteil des Kreises ist das?
    return anteil * kreis_umfang(r), anteil * kreis_flaeche(r)

print(f"1) U = {kreis_umfang(5):.4f}    A = {kreis_flaeche(5):.4f}")
print(f"2) Ringflaeche = {ring_flaeche(5, 3):.4f}")

bogen, sektorflaeche = sektor(6, 72)
print(f"3) Bogenlaenge b = {bogen:.4f}    Sektorflaeche A = {sektorflaeche:.4f}")
```

## Aufgabe 4 – Vielecke: Winkel und Diagonalen

Für ein n-Eck gilt:

* Innenwinkelsumme: `(n − 2) · 180°`
* Innenwinkel im *regelmäßigen* n-Eck: `(n − 2) · 180° / n`
* Anzahl der Diagonalen: `n · (n − 3) / 2`

Erstelle damit eine Tabelle für n = 3 bis 12.

```python
def winkelsumme(n):
    return (n - 2) * 180

def innenwinkel_regelmaessig(n):
    return winkelsumme(n) / n

def diagonalen(n):
    return n * (n - 3) // 2

print(f"{'n':>3} {'Winkelsumme':>13} {'Innenwinkel':>13} {'Diagonalen':>11}")
print("-" * 44)
for n in range(3, 13):
    print(f"{n:>3} {str(winkelsumme(n)) + '°':>13} "
          f"{format(innenwinkel_regelmaessig(n), '.1f') + '°':>13} {diagonalen(n):>11}")
```

## Aufgabe 5 – Flächenformeln und ein Experiment

Zuerst die Standardformeln für Rechteck, Parallelogramm, Trapez und Raute.

Danach ein Experiment: Ein regelmäßiges n-Eck mit Umkreisradius r hat die Fläche

`A = (n/2) · r² · sin(360°/n)`

Was passiert, wenn n immer größer wird? Vermutung aufschreiben, dann laufen lassen.

```python
def rechteck(a, b):
    return a * b

def parallelogramm(g, h):
    return g * h

def trapez(a, c, h):
    return (a + c) / 2 * h

def raute(e, f):
    return e * f / 2

print(f"{'Rechteck (a=6, b=4)':<26} A = {rechteck(6, 4)}")
print(f"{'Parallelogramm (g=6, h=4)':<26} A = {parallelogramm(6, 4)}")
print(f"{'Trapez (a=8, c=4, h=5)':<26} A = {trapez(8, 4, 5)}")
print(f"{'Raute (e=10, f=6)':<26} A = {raute(10, 6)}")

def n_eck_flaeche(n, r):
    return n / 2 * r**2 * math.sin(math.radians(360 / n))

print()
print("Regelmaessiges n-Eck im Kreis mit r = 1:")
for n in [3, 4, 6, 12, 100, 10000]:
    print(f"n = {n:>5}: A = {n_eck_flaeche(n, 1):.8f}")
print(f"Kreis      : A = {math.pi:.8f}   <- genau darauf laeuft es zu")
```
