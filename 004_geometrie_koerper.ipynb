## Aufgabe 1 – Quader und Würfel

Ein Quader hat a = 4, b = 3, c = 12.

1. Volumen `V = a·b·c`
2. Oberfläche `O = 2·(a·b + a·c + b·c)`
3. Raumdiagonale `d = √(a² + b² + c²)` (zweimal Pythagoras hintereinander!)
4. Welche Kantenlänge hätte ein Würfel mit demselben Volumen – und wie viel
   Oberfläche spart er gegenüber dem Quader?

```python
import math

def quader_volumen(a, b, c):
    return a * b * c

def quader_oberflaeche(a, b, c):
    return 2 * (a * b + a * c + b * c)

def raumdiagonale(a, b, c):
    return math.sqrt(a**2 + b**2 + c**2)

a, b, c = 4, 3, 12
V = quader_volumen(a, b, c)
O = quader_oberflaeche(a, b, c)

print(f"1) V = {V}")
print(f"2) O = {O}")
print(f"3) d = {raumdiagonale(a, b, c)}")

kante = V ** (1 / 3)                      # dritte Wurzel aus dem Volumen
O_wuerfel = 6 * kante**2
print(f"4) Wuerfel: Kante = {kante:.4f}, O = {O_wuerfel:.4f}")
print(f"   Ersparnis gegenueber dem Quader: {100 * (1 - O_wuerfel / O):.1f} %")
```

## Aufgabe 2 – Zylinder

Ein Zylinder hat r = 3 und h = 10 (in cm).

1. Volumen `V = π·r²·h`
2. Mantelfläche `M = 2·π·r·h` (das ist das „aufgerollte" Rechteck)
3. Oberfläche `O = 2·π·r² + M`
4. Umgekehrt gefragt: Wie hoch muss ein Zylinder mit r = 3 sein, damit genau
   1 Liter = 1000 cm³ hineinpasst?

```python
def zylinder_volumen(r, h):
    return math.pi * r**2 * h

def zylinder_mantel(r, h):
    return 2 * math.pi * r * h

def zylinder_oberflaeche(r, h):
    return 2 * math.pi * r**2 + zylinder_mantel(r, h)

r, h = 3, 10
print(f"1) V = {zylinder_volumen(r, h):.4f} cm3")
print(f"2) M = {zylinder_mantel(r, h):.4f} cm2")
print(f"3) O = {zylinder_oberflaeche(r, h):.4f} cm2")

# 4) Formel nach h umstellen:  h = V / (pi * r^2)
h_liter = 1000 / (math.pi * r**2)
print(f"4) h = {h_liter:.4f} cm  (Probe: V = {zylinder_volumen(r, h_liter):.4f} cm3)")
```

## Aufgabe 3 – Kegel und Kugel

**Kegel** mit r = 5 und h = 12:

1. Mantellinie `s = √(r² + h²)`
2. Volumen `V = ⅓·π·r²·h`
3. Oberfläche `O = π·r² + π·r·s`

**Kugel** mit r = 6:

4. Volumen `V = 4/3·π·r³`
5. Oberfläche `O = 4·π·r²`

Merkhilfe: Der Kegel passt genau dreimal in den Zylinder mit gleicher Grundfläche
und gleicher Höhe. Das prüfen wir am Ende der Zelle nach.

```python
def kegel_mantellinie(r, h):
    return math.sqrt(r**2 + h**2)

def kegel_volumen(r, h):
    return math.pi * r**2 * h / 3

def kegel_oberflaeche(r, h):
    return math.pi * r**2 + math.pi * r * kegel_mantellinie(r, h)

def kugel_volumen(r):
    return 4 / 3 * math.pi * r**3

def kugel_oberflaeche(r):
    return 4 * math.pi * r**2

r, h = 5, 12
print(f"1) s = {kegel_mantellinie(r, h)}")
print(f"2) V = {kegel_volumen(r, h):.4f}")
print(f"3) O = {kegel_oberflaeche(r, h):.4f}")

print(f"4) Kugel V = {kugel_volumen(6):.4f}")
print(f"5) Kugel O = {kugel_oberflaeche(6):.4f}")

probe = zylinder_volumen(r, h) / kegel_volumen(r, h)
print(f"Probe: Zylinder / Kegel = {probe:.4f}  -> der Kegel passt genau 3x hinein")
```

## Aufgabe 4 – Quadratische Pyramide (Cheops)

Die Cheops-Pyramide hat ungefähr die Grundkante a = 230 m und die Höhe h = 146 m.

1. Volumen `V = ⅓·a²·h`
2. Mantelhöhe (Höhe eines Seitendreiecks) `h_s = √(h² + (a/2)²)`
3. Seitenkante `k = √(h² + 2·(a/2)²)`
4. Oberfläche `O = a² + 2·a·h_s`
5. Wie steil sind die Seitenflächen? `tan(α) = h / (a/2)`

```python
def pyramide_volumen(a, h):
    return a**2 * h / 3

def mantelhoehe(a, h):
    return math.sqrt(h**2 + (a / 2)**2)

def seitenkante(a, h):
    return math.sqrt(h**2 + 2 * (a / 2)**2)

def pyramide_oberflaeche(a, h):
    return a**2 + 2 * a * mantelhoehe(a, h)

a, h = 230, 146
print(f"1) V   = {pyramide_volumen(a, h):.0f} m3")
print(f"2) h_s = {mantelhoehe(a, h):.2f} m")
print(f"3) k   = {seitenkante(a, h):.2f} m")
print(f"4) O   = {pyramide_oberflaeche(a, h):.0f} m2")

neigung = math.degrees(math.atan(h / (a / 2)))
print(f"5) Neigungswinkel der Seitenflaechen: {neigung:.2f} Grad")
```

## Aufgabe 5 – Welcher Körper ist am sparsamsten?

Drei Behälter sollen jeweils genau 1 Liter (1000 cm³) fassen:

* ein Würfel,
* ein Zylinder mit h = 2·r (das ist der materialsparendste Zylinder),
* eine Kugel.

Welcher braucht am wenigsten Oberfläche, also am wenigsten Material?
Erst tippen, dann rechnen lassen.

```python
V_ziel = 1000  # cm3

# Wuerfel:  V = k^3
kante = V_ziel ** (1 / 3)
O_wuerfel = 6 * kante**2

# Zylinder mit h = 2r:  V = pi*r^2*2r = 2*pi*r^3
r_zyl = (V_ziel / (2 * math.pi)) ** (1 / 3)
O_zylinder = zylinder_oberflaeche(r_zyl, 2 * r_zyl)

# Kugel:  V = 4/3*pi*r^3
r_kugel = (3 * V_ziel / (4 * math.pi)) ** (1 / 3)
O_kugel = kugel_oberflaeche(r_kugel)

for name, O in [("Wuerfel", O_wuerfel), ("Zylinder (h=2r)", O_zylinder), ("Kugel", O_kugel)]:
    print(f"{name:<16} O = {O:8.2f} cm2")

print()
print(f"Die Kugel spart gegenueber dem Wuerfel {100 * (1 - O_kugel / O_wuerfel):.1f} % Oberflaeche.")
print("Deshalb sind Seifenblasen rund - die Natur nimmt immer die kleinste Oberflaeche.")
```
